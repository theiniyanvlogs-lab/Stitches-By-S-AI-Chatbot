```python
# ==========================================================
# STEP 3 : Check Folder Structure
# ==========================================================

import os

for root, dirs, files in os.walk(".", topdown=True):
    level = root.replace(".", "").count(os.sep)
    indent = " " * 4 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = " " * 4 * (level + 1)
    for f in files:
        print(f"{subindent}{f}")
```

```python
# ==========================================================
# STEP 5 : Create Data Folders & Move PDFs
# ==========================================================

import os
import shutil

# Create folders
os.makedirs("data/fabric", exist_ok=True)
os.makedirs("data/tailoring", exist_ok=True)
os.makedirs("data/business", exist_ok=True)

# Move uploaded PDFs
for file in os.listdir():
    if file.endswith(".pdf"):

        name = file.lower()

        if "fabric" in name or "cotton" in name:
            shutil.move(file, "data/fabric/" + file)

        elif "tailoring" in name:
            shutil.move(file, "data/tailoring/" + file)

        elif "business" in name:
            shutil.move(file, "data/business/" + file)

print("✅ PDFs moved successfully.")
```

```python
# ==========================================================
# STEP 6 : Verify Uploaded PDFs
# ==========================================================

import os

for folder in [
    "data/fabric",
    "data/tailoring",
    "data/business"
]:
    print("\n", folder)
    print("-"*50)

    files = os.listdir(folder)

    if files:
        for f in files:
            print("✅", f)
    else:
        print("❌ Empty")
```

```python
# ==========================================================
# STEP 7 : Build All Vector Databases
# ==========================================================

import subprocess
import os

scripts = [
    "scripts/build_fabric_db.py",
    "scripts/build_tailoring_db.py",
    "scripts/build_business_db.py"
]

for script in scripts:
    print("\n" + "=" * 70)
    print(f"Running: {script}")
    print("=" * 70)

    result = subprocess.run(
        ["python", script],
        capture_output=True,
        text=True
    )

    print(result.stdout)

    if result.stderr:
        print("ERROR:")
        print(result.stderr)
```

```python
# ==========================================================
# STEP 8 : Verify Vector Databases
# ==========================================================

import os

folders = [
    "vector_db/fabric",
    "vector_db/tailoring",
    "vector_db/business"
]

for folder in folders:
    print("\n" + "="*60)
    print(folder)
    print("="*60)

    if os.path.exists(folder):
        files = os.listdir(folder)

        if files:
            for f in files:
                print("✅", f)
        else:
            print("❌ Empty Folder")
    else:
        print("❌ Folder Not Found")
```
